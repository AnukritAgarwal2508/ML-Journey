```python
#q5
gym['monthly_fee'] = gym['monthly_fee'].str.replace("$", "", regex=False)
gym['monthly_fee'] = pd.to_numeric(gym['monthly_fee'] , errors = 'coerce')
```

```python
#q6
gym['age'] = gym['age'].replace("?", None)
gym['visits_last_month'] = gym['visits_last_month'].replace("--", None)
gym['age'] = pd.to_numeric(gym['age'], errors='coerce')
gym['visits_last_month'] = pd.to_numeric(gym['visits_last_month'], errors='coerce')
```

```python
#q7
gym['join_date'] = pd.to_datetime(gym['join_date'] ,format = "mixed" , dayfirst = True , errors = 'coerce')
```

```python
gym[["monthly_fee", "age", "visits_last_month", "join_date"]].isna().sum()
```

```python
#Q9
gym["age"].dropna().sort_values().tail(10) # bound can be (18 , 120)
gym["visits_last_month"].dropna().sort_values().head(10) # bound will be (0 , 30)
```

```python
#q10
gym.loc[(gym['age']) > 120 , 'age'] = None
gym.loc[(gym['age']) < 18 , 'age'] = None
gym.loc[(gym['visits_last_month']) > 30 , 'visits_last_month'] = None
gym.loc[(gym['visits_last_month']) < 0 , 'visits_last_month'] = None
```

```python
#Q11

gym['plan'] = gym['plan'].replace(
    {
        "STANDARD" : 'standard',
        "Chatswood"   : 'chatswood',
        "Newtown"      : 'newtown',
        "BONDI"  : 'bondi',
        "Parramatta"     : 'parramatta',
        "newtown "        :  'newtown',
        "NEWTOWN"  : 'newtown',
        "CHATSWOOD" : 'chatswood',
        "Bondi"    : 'basic',
        "Basic"    : 'basic'
    }
)
```

```python
#q12

gym['gym'] = gym['gym'].replace(
    {
        "PARRAMATTA" : 'parramatta'
    }
)
```
