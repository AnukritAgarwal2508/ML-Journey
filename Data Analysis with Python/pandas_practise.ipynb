```python
import pandas as pd

data = [1,2,3,4,5]
series_from_list = pd.Series(data)
print(series_from_list)


data = {'a' : 1 , 'b' : 2 , 'c' : 3}
series_from_dictionary = pd.Series(data)
print(series_from_dictionary)


data = [[1,2,3],[4,5,6],[7,8,9]]
idx = ['a','b','c']

series_custom_index = pd.Series(data,index = idx)
print(series_custom_index)


### indexing in series

print(series_from_list[0])
print(series_custom_index['a'])

```

```python
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([6,7,8,9,10])

sum = s1 + s2
print(sum)
```

```python
series_a = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
series_b = pd.Series([4, 5, 6], index=['b', 'c', 'd'])
sum_series = series_a + series_b 
print(sum_series)
```
