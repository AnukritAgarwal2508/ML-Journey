```python
df2["salary_band"] = np.select(
    [df2["salary"] < 75000, df2["salary"] < 110000],
    ["low",                  "mid"                 ],
    default="high"
)

df2["salary_band"].value_counts()
```

```python
df2["salary_band2"] = pd.cut(
    df2["salary"],
    bins=[0, 75000, 110000, np.inf],
    labels=["low", "mid", "high"]
)

df2["salary_band2"].value_counts()
```
