```python
df['date_of_joining'] = pd.to_datetime(
    df["date_of_joining"],
    ## in the data there are various formats of data so to change them to a proper datetime format , format = "mixed" is required
    format = "mixed",
    ## days should comes first the format should be dd/mm/yyyy
    dayfirst = True,
    ## without this any unparsed value can cause an error in the code
    errors = "coerce"
)
```

```python
df['date_of_joining'].describe()
```

```python
df['date_of_joining'].isna().sum()
```

```python
df['tenure_years'] = (pd.Timestamp("today") - df["date_of_joining"]).dt.days / 365.25

## pd.timestamp("todays") -> takes todays date
## dt.days -> converts the result into days
#/365.25 -> convert the number of days to years
```
