```python
df["department"].replace({"mktg": "marketing", "h.r.": "hr", "human resources": "hr", "engg": "engineering"})
df["department"].str.strip().str.lower().value_counts()
```

```python
df["department"] = (
    df["department"]
      .str.strip()
      .str.lower()
      .replace({
          "mktg": "marketing",
          "h.r.": "hr",
          "human resources": "hr",
          "engg": "engineering",
          "customer support": "support",
          "?": None,
      })
)

df["department"].value_counts(dropna=False)
```

```python
df['city'].value_counts()

## the output is giving 21 different names for 6 cities we have to clean it 

df['city'] = df['city'].str.strip().str.lower()

## the city is now fixed :)
df['city'].value_counts()

```

```python
s = df['salary']

s = s.str.replace("$","",regex = False)
s = s.str.replace(",","",regex = False)
s = s.str.replace(" AUD","",regex = False)

df['salary'] = s

df['salary'] = pd.to_numeric(s , errors = "coerce")


```

```python
df['salary'] = df['salary'].abs()

df['salary'].describe()
```
