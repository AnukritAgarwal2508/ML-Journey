```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('cafe_sales.csv')
bondi = df[df["store"] == "Bondi"]

first_30 = bondi.head(30)
plt.plot(first_30['date'] , first_30["revenue"])
plt.show()
```

```python
week = bondi.head(7)
plt.bar(week["day_of_week"] , week["revenue"])
plt.show()
```

```python
plt.scatter(df["temp_c"] , df["hot_drinks"])
plt.show()
```

```python
plt.hist(df["temp_c"])
plt.show()
```

```python
df.corr(numeric_only = True)
```

```python
plt.figure(figsize = (10 , 5))
plt.plot(first_30['date'] , first_30['revenue'])
plt.title("Bondi daily revenue, first 30 days")
plt.xlabel('Date')
plt.ylabel('Revenue (AUD)"')
plt.xticks(rotation = 45)
plt.grid(True)
plt.show()
```

```python
df.corr(numeric_only = True)
```

```python
plt.figure(figsize=(10, 4))
plt.plot(first_30["date"], first_30["revenue"])
plt.ylim(0, 1600)
plt.title("Starts at zero")
plt.xticks(rotation = 45)
plt.show()

plt.figure(figsize=(10, 4))
plt.plot(first_30["date"], first_30["revenue"])
plt.ylim(900, 1400)
plt.xticks(rotation = 45)
plt.title("Starts at 900")
plt.show()
```

```python
newtown = df[df["store"] == "Newtown"]

plt.figure(figsize=(10, 5))
plt.plot(bondi.head(30)["date"], bondi.head(30)["revenue"], label="Bondi")
plt.plot(newtown.head(30)["date"], newtown.head(30)["revenue"], label="Newtown")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```
