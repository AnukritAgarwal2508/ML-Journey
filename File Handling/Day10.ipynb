```python
with open("demo.bin", "wb") as f:
    f.write(bytes([104, 105, 200]))     # container in, file created

with open("demo.bin", "rb") as f:
    a = f.read(1)        # b'h'      finger 0 -> 1
    b = f.read(1)        # b'i'      finger 1 -> 2
    c = f.read(1)        # b'\xc8'   200 has no letter
    d = f.read(1)        # b''       nothing left

print(list(a), a[0])     # [104] 104   container vs number
print(a.decode())        # h           numbers -> text
```

```python
#in order to write a text inside a binary file do this

text = "hello"

with open("out.bin","wb") as file:
    file.write(text.encode())

with open("out.bin","rb") as file:
    second = file.read(2)

print(second)
```

```python
with open("data.bin", "wb") as f:
    f.write(bytes([10, 20, 30, 40, 50]))

with open("data.bin", "rb") as f:
    while True:
        chunk = f.read(2)
        if chunk == b"":
            break
        print(list(chunk))
```

```python
### read the content and write it to new file 

with open('example.txt','r') as source_file:
    content = source_file.read()

### if file is not there it will automatically create the file
with open('destination.txt' ,'w') as destination_file:
    destination_file.write(content)

```

```python
#### writing and then reading a file

with open('example.txt' , 'w+') as file:
    file.write("HELLO WORLD\n")
    file.write("This is a new line \n")

    ## move the file cursor to the begginign
    file.seek(0) 
    
    content = file.read()
    print(content)
```
