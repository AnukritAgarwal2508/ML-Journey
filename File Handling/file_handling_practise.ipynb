```python
word_count = 0
with open('output.txt','r') as file:
    for line in file:
        line = line.split()
        for words in line:
            if words == ' ':
                continue
            else:
                word_count += 1

print(word_count)
```

```python
lines = 0
words = 0
characters = 0

with open('output.txt','r') as file:
    for line in file:
        line = line.split()
        lines += 1
        for word in line:
            if word != ' ':
                words += 1
            
            for characer in word:
                    characters += 1

print(lines , words , characters)
```
