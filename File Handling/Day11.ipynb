```python
## checking if a file is either directiory or file
import os
path = 'example1.txt'

if os.path.isfile(path):
    print("This is a file")

elif os.path.isdir(path):
    print("this is a directory")

else:
    print("this is neither file nor directory")
```

```python
### getting the abs path

relative_path = 'example1.txt'

absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```
