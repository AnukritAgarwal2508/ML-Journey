```python
class Book:
    def __init__(self, title, author, rate):
        self.title = title
        self.author = author
        self.rate = rate   # goes through the setter now

    @property
    def rate(self):
        return self.__rate

    @rate.setter
    def rate(self, rate):
        if rate < 0:
            return 'not a valid rate'
        else:
            self.__rate = rate


class Ebook(Book):
    def __init__(self, title, author, rate, file_size):
        super().__init__(title, author, rate)
        self.file_size = file_size


e1 = Ebook('The tale of christmas', 'Anukrit', 78, 10.5)
e2 = Ebook('The tale of Ammakalam vinay', 'Aditya', -9, 20.3)

print(e1.author, e1.title, e1.rate, e1.file_size)
print(e2.author, e2.title, e2.rate, e2.file_size)
```

```python
class Product:

    def __init__(self , name , price = None) -> None:
        self.name = name
        self.__price = price

    
    def get_price(self):
        return self.__price
    

    def set_price(self , price):
        if price < 0:
            print("price cant be negative pls change the price value")
        
        else:
            self.__price = price


p1 = Product('Asprin')
p1.set_price(-10)
p2 = Product('Gamma')
p2.set_price(50)

print(p1.get_price())
print(p2.get_price())

```

```python
from abc import ABC , abstractclassmethod , abstractmethod

class SchoolMember(ABC):

    @abstractmethod
    def role(self):
        pass

class Person:

    def __init__(self , name) -> None:
        self.name = name
    

    def __str__(self) -> str:
        return f'{self.name}'
    
    def __repr__(self) -> str:
        return f'Person({self.name})'
    
    def describe(self):
        return f'{self.name}'

class Student(SchoolMember , Person):

    def __init__(self, name , marks) -> None:
        super().__init__(name)
        self.marks = marks
    
    def __str__(self) -> str:
        return f'{self.name} , {self.marks}'
    
    def __repr__(self) -> str:
        return f'Student({self.name} , {self.marks})'

    @property
    def marks(self):
        return self.__marks
    
    @marks.setter
    def marks(self , marks):
        if marks < 0 or marks > 100:
            raise ValueError('Invalid Marks')
        
        else:
            self.__marks = marks
    
    def grade(self):
        if self.marks >= 85:
            return 'A'
        
        if self.marks >= 60:
            return 'B'
        if self.marks >= 40:
            return 'C'
        
        else:
            return 'F'

    def add_bonus(self , bonus):
        if self.__marks + bonus > 100 or bonus < 0:
            raise ValueError('Invalid Bonus')
        
        self.__marks += bonus

        print(f"{self.name} got {bonus} marks as bonus")
    
    def isPass(self):
        if self.marks >= 40:
            return True
        
        else:
            return False
    
    def describe(self):
        super().describe()
        print(f'scored {self.__marks} marks')
    
    def role(self):
        return 'Student'
    

    def __add__(self , other):
        return self.__marks + other.__marks
    

    def __lt__(self , other):
        return self.__marks < other.__marks


class Teacher(SchoolMember):

    def __init__(self , name):
        self.name = name

    def role(self):
        return 'Teacher'



class Library_Member:

    def borrow(self , book):
        print(f"{self.name} borrowed {book}")
    

class GradStudent(Student , Library_Member):

    def __init__(self, name, marks , thesis_topic) -> None:
        super().__init__(name, marks)
        self.thesis_topic = thesis_topic    
    
    def __str__(self) -> str:
        return f'{self.name} , {self.marks} , {self.thesis_topic}'
    
    def __repr__(self) -> str:
        return f'GradStudent({self.name} , {self.marks} , {self.thesis_topic})'
    
    def describe(self):
        super().describe()
        print(f'researching on {self.thesis_topic}')

class Classroom:

    def __init__(self) -> None:
        self.students = []
    

    def add(self , student):
        self.students.append(student)

    
    def average_marks(self):
        marks = 0
        for i in range(len(self.students)):
            marks += self.students[i].marks
        
        return marks//len(self.students)


try:
    m = SchoolMember()
except TypeError as e:
    print("Error:", e)

s1 = Student("Aditya", 78)
s2 = Student("Anukrit", 92)
print(s1 + s2)
print(s1 < s2)
print(s1.role(), Teacher("Ms. Rao").role())

room = Classroom()
room.add(s1)
room.add(s2)
print(room.average_marks())





        
```

```python
class BankAccount:

    def __init__(self , owner , balance) -> None:
        self.owner = owner
        self.__balance = balance

    
    def get_balance(self):
        return self.__balance
    
    def deposit(self ,amount):
        if amount <= 0:
            raise ValueError('Deposit amount cant be -ve or 0')
        
        else:
            self.__balance += amount

    def withdraw(self , amount):
        if amount > self.__balance:
            raise ValueError('Insufficient Balance')
        
        if amount <= 0:
            raise ValueError('Invalid amount')
        
        else:
            self.__balance -= amount


acc1 = BankAccount('Anukrit' , 1000)
print (f" Current Balance : {acc1.get_balance()}")

acc1.deposit(980)
print (f" Current Balance : {acc1.get_balance()}")

acc1.withdraw(32)
print (f" Current Balance : {acc1.get_balance()}")

acc2 = BankAccount('Aditya' , 900)
print (f" Current Balance : {acc2.get_balance()}")

try:
    acc2.deposit(-89)

except ValueError as e:
    print("Error : ",e)

print (f" Current Balance : {acc2.get_balance()}")

try:
    acc2.withdraw(-10000)

except ValueError as e:
    print("Error : ",e)

print (f" Current Balance : {acc2.get_balance()}")


```
