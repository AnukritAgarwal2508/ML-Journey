```python
class Animal:

    def __init__(self , name , species) -> None:
        
        self.name = name
        self.species = species
    
    
class Dog(Animal):

    def __init__(self, name, species , breed) -> None:
        super().__init__(name, species)
        self.breed = breed
    
    @property
    def bark(self):

        print(f"{self.name} says woof")

    def __str__(self):
        
        return f"Dog info -> Name : {self.name} , Species : {self.species} , Breed : {self.breed} "

dog1 = Dog("pluto" , "Anthropoda" , "Labrador")
dog1.bark
print(dog1)
```

```python
class Walker:

    @property
    def Walk(self):

        print("Its time to walk")
    
class Run:

    @property
    def Run(self):

        print("Its time to run")

class Athelete(Walker , Run):

    def __init__(self , training_hours , train) -> None:
        Walker.Walk
        self.training_hours = training_hours
        self.train = train

    def Walk(self):
        
        print("Athelete must walk")



ath1 = Athelete()
ath1.Walk
ath1.Run
```

```python
class A:

    def show(self):

        print("Message")
    

class B(A):

    def show(self):

        print("Message_b")


class C(A):

    def show(self):

        print("Message_c")


class D(B , C):

    pass

d_obj = D()
d_obj.show()
d_obj.show()


```

```python
class Person:

    def __init__(self , name) -> None:
        self.name = name

class Employee:

    def __init__(self , employee_id) -> None:
        self.employee_id = employee_id

class Manager(Person , Employee):

    def __init__(self , name , employee_id) -> None:
        super().__init__(name) #-> sends the name variable to the Persons's init 
        Employee.__init__(self , employee_id)   


m1 = Manager("Anukrit", 12345)



```

```python
class Base1:

    def __init__(self , a) -> None:
        self.a = a
        print(f"A is initalized : {self.a}")

class Base2:

    def __init__(self , b) -> None:
        self.b = b
        print(f"B is initalized : {self.b}")

class Base3(Base1 , Base2):

    def __init__(self, a , b) -> None:
        super().__init__(a)
        Base2.__init__(self , b)


b3 = Base3(1,2)
```

```python
class Bird:

    @property
    def speak(self):
        print("Bird Speaks")

class Parrot(Bird):

    @property
    def speak(self):
        print("Parrot speaks")


class Penguin(Bird):

    @property
    def speak(self):
        print("Penguin speaks")

bird1 = Bird()
bird2 = Parrot()
bird3 = Penguin()

bird1.speak
bird2.speak
bird3.speak
```

```python
class Device:

    def __init__(self, brand) -> None:
        self.brand = brand

class Phone(Device):

    def __init__(self, brand , model) -> None:
        super().__init__(brand)
        self.model = model

class Camera:

    def __init__(self , resolution) -> None:
        self.resolution = resolution

class Smartphone(Phone , Camera):

    def __init__(self, brand, model , resolution) -> None:
        super().__init__(brand, model)
        Camera.__init__(self , resolution)

sm1 = Smartphone("Samsung" , "I8790" , "4K")
print(sm1.brand)
print(sm1.model)
print(sm1.resolution)


        
```
