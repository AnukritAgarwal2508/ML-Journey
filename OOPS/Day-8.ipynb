```python
class Animal:

    def __init__(self , name):
        self.name = name
    

    def speak(self):
        print("Subclass must implement this method")


class Pet:

    def __init__(self , owner):
        self.owner = owner
    
    def cuteness(self):
        print(f"{self.owner} has a very cute pet")


class dog(Animal , Pet):

    def __init__(self, name , owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    
    def speak(self):
        print(f"{self.name} of {self.owner} speaks Woof")


dog1 = dog("Wolfie" , "Anukrit")
dog1.speak()
dog1.cuteness()
```

```python
class Book:

    def __init__(self , name , publishing_year , author , rating) -> None:
        self.name = name
        self.publishing_year = publishing_year
        self.author = author
        self.rating = rating

    
    def book_descripion(self):
        print(f"{self.name} is having {self.rating} rating")


class Comic_Book(Book):

    def __init__(self, name, publishing_year, author, rating , no_of_pictures) -> None:
        super().__init__(name, publishing_year, author, rating)
        self.no_of_pictures = no_of_pictures
    
    def show_info(self):
        print(f"{self.name} has {self.no_of_pictures} pictures")


class DC_Comic(Comic_Book):

    def __init__(self, name, publishing_year, author, rating, no_of_pictures , hero_name) -> None:
        super().__init__(name, publishing_year, author, rating, no_of_pictures)
        self.hero_name = hero_name
    
    def which_hero(self):
        print(f"{self.name} is about {self.hero_name}")


comic1 = DC_Comic('Far from the sea' , 1997 , "James Bond" , "Excellent" , 45 , 'Stardust')

comic1.book_descripion()
comic1.show_info()
comic1.which_hero()
```

```python
## Polymorphism

### Method Overiding

class Animal:
    def speak(self):
        return "Sound of the Animal"

# Derived Class 1
    
class Dog(Animal):
    def speak(self):
        return "Woof"

#Derived Class
    
class Cat(Animal):
    
    def speak(self):
        return "Meow"

#animal_speak function is recieving animal as an object
def animal_speak(animal):
    print(animal.speak())


dog = Dog()
cat = Cat()

print(dog.speak())
print(cat.speak())
animal_speak(dog)
```

```python
## Polymorphism with Functions and methods

class Shape:
    def area(self):
        return "The area of the figure"

class Rectangle(Shape):
    def __init__(self , width , height) -> None:
        self.width = width
        self.height = height
    
    def area(self):
        return f"{self.width * self.height} is the area"


class Circle(Shape):
    def __init__(self , radius) -> None:
        self.radius = radius
    
    def area(self):
        return f"{3.14 * self.radius * self.radius} is the area"
    

def area(shape):
    return shape.area()


```

```python
## Polymorphisme with abstract base class 

from abc import ABC , abstractmethod

## define an abstract class

class Vechile(ABC):
    @abstractmethod

    def start_engine(self):
        pass

class Car(Vechile):

    def start_engine(self):
        return "Car Engine started"
    

class Truck(Vechile):

    def start_engine(self):
        return "Truck Engine started"


def engine(vehicle):
    return vehicle.start_engine()

car1 = Car()
truck1 = Truck()

print(engine(car1))
print(engine(truck1))
```

```python
## Encapsulation with getter and setter 

class Person():

    def __init__(self , name , age):
        self.__name = name ## private
        self.__age = age
    
    ##getter methof for name
    def get_name(self):
        return self.__name
    
    ## setter method
    def set_name(self , name):
        self.__name = name
    
    def get_age(self):
        return self.__age

    def sett_age(self , age):
        if age > 0:
            self.__age = age
        
        else:
            print("Age cant be negative")

person = Person("SANYA" , 32)
print(person.get_name())
print(person.get_age())

person.set_name("Anukrit")
print(person.get_name())

person.sett_age(-5)
print(person.sett_age())
            
```

```python
from abc import ABC , abstractmethod

##HERE vehicle is an abstract class ABC means that the class is of Abstract type
class Vechicle(ABC):
    def drive(self):
        print("The vehicle is used for driving")
    
    @abstractmethod
    #if a child class is inhering the parent class excxept the abstract method 
    #child class can use anything ofthe parent class
    #however in order to use the abstract method it have to define its own way of using it
    def start_engine(self):
        pass


class Car(Vechicle):

    def start_engine(self):
        print("Car engine started")


def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()


car = Car()
operate_vehicle(car)
    
```

```python
## operatior overloading

class Vector:

    def __init__(self , x , y) -> None:
        self.x = x
        self.y = y
    
    def __add__(self , other):
        return Vector(self.x + other.x , self.y + other.y)
    

    def __mul__(self , other):
        return Vector(self.x * other.x , self.y * other.y)
    
    def __sub__(self , other):
        return Vector(self.x -  other.x , self.y - other.y)
    
    def __eq__(self , other):
        return self.x == other.x and self.y == other.y

    def __repr__(self) -> str:
        return f"Vector({self.x}, {self.y})"    

## creat objects of the vector clas
```

```python
## another example of operator overloading

class Complex_Number:

    def __init__(self, real_1, imag_1):
        self.real_1 = real_1
        self.imag_1 = imag_1

    def __add__(self, other):
        return Complex_Number(self.real_1 + other.real_1, self.imag_1 + other.imag_1)

    def __sub__(self, other):
        return Complex_Number(self.real_1 - other.real_1, self.imag_1 - other.imag_1)

    def __mul__(self, other):
        return Complex_Number(
            self.real_1 * other.real_1 - self.imag_1 * other.imag_1,
            self.real_1 * other.imag_1 + self.imag_1 * other.real_1
        )

    def __truediv__(self, other):
        return "no such operation in complex numbers"

    def __repr__(self):
        return f"({self.real_1} + {self.imag_1}i)"
    

c1 = Complex_Number(3 , 4)
c2 = Complex_Number(5 , 6)

print(c1 + c2)
print(c1 - c2)
print(c1 * c2)
print(c1 / c2)



```

```python
# Class Composition 

class Engine:

    def start(self):
        print("Engine is starting")

class Car():

    ## car is holding the class Engine that is it is an "has A relationship" with another class not "is a "
    def __init__(self) -> None:
        self.engine = Engine()

my_car = Car()
my_car.engine.start()
```
