```python
class Error(Exception):
    pass

class DobException(Error):
    pass

year = int(input("Enter the DOB : "))
age = 2024 - year

try:
    if age <=30 and age >= 20:
        print("Valid")

    else:
        raise DobException

except DobException:
    print("Sorry your age does not follow the rules")
```
