```python
class Car:

    def __init__(self , make , model , year) -> None:

        self.make = make
        self.model = model
        self.year = year
    
    def details(self):
        print(f"Car Details : {self.make} , {self.model} , {self.year}")
    
    def engine_start(self):
        print(f"{self.make}'s engine has started")


car1 = Car("Audi" , "IGYT32" , 1997)
car1.details()
car1.engine_start()
```

```python
class BankAccount:

    def __init__(self , account_number , balance) -> None:

        self.__account_number = account_number
        self.__balance = balance
    

    def deposit(self , money):
        print(f"Money Successfull Deposited , Avalible Balance : {self.__balance + money}")
    

    def withdraw(self , money):
        if money > self.__balance:
            print("Insufficient Balance")
        
        else:
            print(f"Money Succesfully Withdrawn , , Avalible Balance : {self.__balance - money}")

acc1 = BankAccount(1243562909 , 80000)

acc1.deposit(233)
acc1.withdraw(432)
acc1.withdraw(100000)
```

```python
## static method - method belonging to a class instead of object
import math

class Area:

    def __init__(self, number):
        self.number = number
    
    def root(self):
        return math.pow(self.number , 1/2)
    


even = Area(19)
print(even.root())
        
```

```python
class Rectangle:

    def __init__(self , length , width):
        self.__length = length
        self.__width = width
    
    
    ##getter method
    @property
    def length(self):
        return self.__length
    
    ##setter for length
    @length.setter
    def length(self , new_length):
        if new_length < 0:
            print("length cant be negative")
        
        else:
            self.__length = new_length
            print("length has changed")

    
            
    ##getter method
    @property
    def width(self):
        return self.__width
    
    ##setter for length
    @width.setter
    def width(self , new_width):
        if new_width < 0:
            print("width cant be negative")
        
        else:
            self.__width = new_width
            print("width has changed")
    
        


rect1 = Rectangle(34 , 21)
print(rect1.length)
rect1.width = 45
print(rect1.width)
```

```python
from abc import ABC , abstractclassmethod

class Shape(ABC):

    @abstractclassmethod
    def area(self):
        pass


class Circle(Shape):

    def __init__(self , radius) -> None:
        self.radius = radius
    
    @property
    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):

    def __init__(self , length , width):
        self.length = length
        self.width = width
    
    @property
    def area(self):
        return self.length * self.width
    

cicr1 = Circle(7)
print(cicr1.area)

rect1 = Rectangle(12 , 23)
print(rect1.area)
```

```python
class Vector:

    def __init__(self , x , y):

        self.x = x
        self.y = y
    
    def __add__(self , other):

        return Vector(self.x + other.x , self.y + other.y)
    
    def __subtract__(self , other):

        return Vector(self.x - other.x , self.y - other.y)


Vec1 = Vector(3 , 4)
Vec2 = Vector(5 , 2)

Vec3 = Vec1 + Vec2
print(Vec3.x)
print(Vec3.y)
```

```python
class Vector:

    def __init__(self , first_no , second_no):

        self.first_no = first_no
        self.second_no = second_no
    
    def __add__(self , other):

        return Vector(self.first_no + other.first_no , self.second_no + other.second_no)
    
    def __sub__(self , other):

        return Vector(self.first_no - other.first_no , self.second_no - other.second_no)
    
    def __repr__(self):

        return f"Vector({self.first_no} , {self.second_no})"
    

V1 = Vector(5 , 6)
V2 = Vector(3 , 4)

print(V1 + V2)

    
```
