```python
class Rectangle:

    def __init__(self , length , width):
        self.__length = length
        self.__width = width
    
    
    ##getter method
    def get_length(self):
        return self.__length
    
    def set_width(self , new_width):
        if new_width < 0:
            print("Width cant be negative")
        
        else:
            self.__width = new_width
            print("Width has changed")

rect1 = Rectangle(34 , 21)
print(rect1.get_length())

rect1.set_width(45)
```

```python
class Rectangle:

    def __init__(self , length , width):
        self.__length = length
        self.__width = width
    
    
    ##getter method
    @property
    def length(self):
        return self.__length
    
    ##setter for length
    @length.setter
    def length(self , new_length):
        if new_length < 0:
            print("length cant be negative")
        
        else:
            self.__length = new_length
            print("length has changed")

    
            
    ##getter method
    @property
    def width(self):
        return self.__width
    
    ##setter for length
    @width.setter
    def width(self , new_width):
        if new_width < 0:
            print("width cant be negative")
        
        else:
            self.__width = new_width
            print("width has changed")
    
        


rect1 = Rectangle(34 , 21)
print(rect1.length)
rect1.width = 45
print(rect1.width)
```
