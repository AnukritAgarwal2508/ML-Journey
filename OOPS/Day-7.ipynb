```python
class BankAccount:
    def __init__(self , owner , balance = 0):
        self.owner = owner
        self.balance = balance
    
    def deposit(self , balance):
        self.balance += balance
        print(f"The money has been successfully deposited -> Current Amount - {self.balance}")
    
    def withdraw(self , balance):
        if balance > self.balance:
            print("Insufficient Balance")
        
        else:
            self.balance -= balance
            print(f"The money has been successfully withdrawed -> Current Amount - {self.balance}")

person1 = BankAccount("Anukrit" , 300000)
person2 = BankAccount("Arnav" , 40000)

person1.deposit(700000)
person1.withdraw(20000000)

person2.deposit(7888)
person2.withdraw(4550)
```

```python
## inheritance

class Car_inheritance:
    def __init__(self , windows , doors , engine_type):

        self.windows = windows
        self.doors = doors
        self.engine_type = engine_type
    
    def drive(self):
        print(f"The person will drive the {self.engine_type} car")
    

car1 = Car_inheritance(4,5,"Petrol")
car1.drive()
## this is where the inheritance is happening

class Tesla(Car_inheritance):
    def __init__(self, windows, doors, engine_type, is_self_driving):
        super().__init__(windows, doors, engine_type)
        self.is_self_driving = is_self_driving

    def self_driving(self):
        print(f"Tesla supports self driving: {self.is_self_driving}")


tesla1 = Tesla(4 , 5 , "Electric" , True)
tesla1.self_driving()

tesla1.drive()



```

```python
class Employee:

    def __init__(self , name , salary):
        self.name = name
        self.salary = salary
    
    def work(self):
        print(f"{self.name} is working")

class Manager(Employee):

    def __init__(self , name , salary , team_size):
        super().__init__(name , salary)
        self.team_size = team_size
    

    def manage(self):
        print(f"{self.name} manages a team of {self.team_size}")

m1 = Manager("Asha", 90000, 6)
m1.work()     # should still work, even though Manager never defines it
m1.manage()   # the new behavior only Manager has
```

```python
class Shape:

    def __init__(self , name):
        self.name = name
    
    def area(self):
        return 0
    
    def describe(self):
        print(f"This is a {self.name} with area {self.area()}")

class Rectangle(Shape):

    def __init__(self,name , width , height):
        super().__init__(name)
        self.width = width
        self.height = height

    
    def area(self):
        return self.width * self.height


r1 = Rectangle("Rectangle " , 4, 5)
r1.describe()
    

```
