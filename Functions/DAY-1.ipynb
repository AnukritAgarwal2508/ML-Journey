```python
def greet():
    print("Hello Anukrit")

#parameter with a name
def greet_new(name):
    print("Hello",name)

def greet_multiple(name1,name2):
    ''' Prints the name of two people as per the arugement 
    This is the concept of Docstring - tells what the function do'''

    print("Hello",name1,"and",name2)

def greet_default(name = "Guest"):
    print(name)

def greet_variable(*args):
    '''Stores Multiple Arguments inside a tuple'''
    for name in args:
        print(name)

def info_of_person(**kwargs):
    ''' stores multiple arguments and their values inside a dictionary'''

    for key,values in kwargs.items():
        print(key,":",values)


greet()
greet_new("Aditya")
greet_multiple("Anukrit","Aditya")
greet_default()
greet_default("Anukrit")
greet_variable("Anu","Aditya","Sreesh","Hemanth")
greet_variable("Anu","Aditya")

info_of_person(name = "Anukrit" , age = "21" , gender = "Male")




```

```python
def multiple_arguments(*args,**kwargs):
    ''' we can use both args and kwargs but args should come before kwargs'''

    print(args) # will print values stored inside a tuple
    print(kwargs) # will print values stored inside a dictionary


multiple_arguments(1,2,3,name = "Anukrit",age = "21")
```

```python
def temprature_converter(temprature,unit):
    if unit == 'F':
        new_temprature = (temprature * 9/5) + 32
        return new_temprature
    
    new_temprature = (temprature - 32) * 5 / 9
    return new_temprature


new_temprature = temprature_converter(68,'C')
print(new_temprature)
```

```python
def is_password_strong(password):
    '''checks if password is strong or not'''
    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    
    return True

## calling the function
print(is_password_strong("12345"))
print(is_password_strong("ANuman@0803"))
```

```python
def calculate_total_cost(cart):
    total_cost = 0
    print("Bill")
    for item in cart:
        total_cost += item['cost'] * item['quantity']
        print("Item : ",item['name'] , " , Total_cost : ",item['cost'] * item['quantity'])
    
    return total_cost

cart = [
    {'name':"Apple",'cost':0.5,'quantity':4},
    {'name':"Banana",'cost':0.6,'quantity':6},
    {'name':"Orange",'cost':0.7,'quantity':3}

]
## calling the function
total_cost = calculate_total_cost(cart)
print("Your total bill is : ",total_cost)
```

```python
def is_palindrome(s):
    ''' check is function is palindrome or not'''
    s = s.lower().replace(" ","")
    return s[::-1] == s

print(is_palindrome("Anukrit"))
print(is_palindrome("A man a plan a canal Panama"))


```

```python
def factorial(number):
    if number == 0 or number == 1:
        return 1
    
    return number * factorial(number - 1)

print(factorial(3))
```

```python
def count_word_frequency(file_path):
    word_count = {}
    with open(file_path,'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.lower().strip('•,!?;:"\'')
                word_count[word] = word_count.get(word,0) + 1
    
    return word_count

file_path = 'sample.txt'
word_frequency = count_word_frequency(file_path)
print(word_frequency)
```
