```python
def fibonacci_sequence(n):
    if n < 0 :
        print("Invalid Number")
        return 
    if n == 0 or n == 1:
        return n
    
    return fibonacci_sequence(n - 1) + fibonacci_sequence(n - 2)

print(fibonacci_sequence(3))
```

```python
def only_integers(**kwargs):
    spare_dict = {}
    for key,value in kwargs.items():
        if isinstance(value,int):
            spare_dict[key] = value
        
    
    return spare_dict


new_dict = only_integers(name = 'Anukrit' , age = 21 , Background = 'CS' , Job_in_lpa = 80 , no_of_friends = 5)
print(new_dict)
```

```python
def callback_function(func_name , list_of_numbers):
    ''' function takes another function and implement on every number on the list
    concept of map and lambda can be used here'''
    return list(map(func_name,list_of_numbers))


square = lambda x : x ** 2 
double = lambda x : x * 2
to_string = lambda x : str(x)
numbers = [1,2,3,4,5]

print(callback_function(square,numbers))
print(callback_function(double,numbers))
print(callback_function(to_string,numbers))

```

```python
def higher_order_function(numbers):

    new_list = filter_function(numbers)
    return map_function(new_list)

def filter_function(list_of_numbers):
    return list(filter(lambda x : x % 3 == 0,list_of_numbers))

def map_function(updated_list_of_numbers):
    return list(map(lambda x : x ** 2 , updated_list_of_numbers))


numbers = [x for x in range(1 , 100)]
print(higher_order_function(numbers))

```

```python
def func_of_func(func1 , func2 , number):
    ''' performs f(g(x) where f and g are two function'''
    return list(map(func2,list(map(func1,number))))


func1 = lambda x : x ** 2
func2 = lambda x : str(x)

number = [x for x in range(1 , 10)]
print(func_of_func(func1,func2,number))
```

```python
def mini_operation_calculator(operation):
    ''' function inside function can be used to return function or whatever operation we need to perform on the data'''
    if operation == '+':
        def add(a , b):
            return  a + b
        
        return add
    
    elif operation == '-':
        def subtract(a , b):
            return a - b
        
        return subtract
    
    elif operation == '*':
        def multiplication(a , b):
            return a * b
        
        return multiplication
    
    else:
        def divide(a , b):
            if b == 0:
                return None
            
            return a/b
        return divide
    

operation = mini_operation_calculator('+')
print(operation(2 , 3))
operation = mini_operation_calculator('-')
print(operation(2 , 3))
operation = mini_operation_calculator('*')
print(operation(2 , 3))
operation = mini_operation_calculator('/')
print(operation(2 , 3))



```

```python
from functools import partial

def add(n1,n2,n3,n4):
    return n1+n2+n3+n4


new_add = partial(add,2,3) ## creates a new function here now n1 = 2 , n2 = 3
print(new_add(6,7))

''' will give an error because 2 arguements of the function are already occupied
so error will come - giving 6 positional arguments instead of 4'''
print(new_add(1,2,3,4))

''' during the partial function it is neccessary that passing same type of arguments during partial function 
defining and calling ie either the arguments should be positional ie ( 2, 3 )
or argumental ie(n1 = 2 , n2 = 3)'''

```

```python
def first_generator():
    print("Start")
    yield 1
    print("Middle")
    yield 2
    print("End")
    yield 3

gen = first_generator()
## will not return the value instead it will return generator  object
print(next(gen)) ## the function will not terminate instead it will start from the lines which lies ahead of yield
print(next(gen))
print(next(gen))
## no more yields left so it will give an error
print(next(gen))
```
