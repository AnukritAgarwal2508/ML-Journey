```python
def mse(y_actual , y_pred):
    square = 0

    if len(y_pred) != len(y_actual):
        raise ValueError(
            f"Length mismatch: {len(y_actual)} true values vs {len(y_pred)} predictions"
        )
    
    if len(y_actual) == 0:
        raise ValueError(
            f'Cannot compute mse on the empth list'
        )
    
    for pred , actual in zip(y_pred , y_actual):
        diff = pred - actual
        square += diff ** 2
    
    return square/len(y_pred)


try:
    mse([1, 2, 3], [1, 2])
except ValueError as e:
    print("Caught:", e)


try:
    mse([], [])
except ValueError as e:
    print("Caught:", e)



```
