```python
def fileopen(file_name):
    try:
        with open(file_name, 'r') as file:
            data = file.read()
            return data
    except FileNotFoundError as e:
        return f"Message : {e}"


print(fileopen('data.txt'))
```

```python
class new_class:

    def __init__(self , num1 , num2):
        self.num1 = num1
        self.num2 = num2
    
    def division(self):
        try:
            result = self.num1/self.num2
            return result
        except ZeroDivisionError as e:
            print("Zero division not allowed")
            return "Operation Terminated"
    

n1 = new_class(9 , 9)
n2 = new_class(9 , 0)

print(n1.division())
print(n2.division())
```

```python
def convert_string_to_int(lst):
    try:
        lst = [int(i) for i in lst]
        return lst
    except ValueError:
        print("Error")

lst = ["1" , "2" , "3" , "4" , "abc"]
convert_string_to_int(lst)
```

```python
def file_writing(file_name , lst):
    try:
        with open(file_name , 'w') as file:
            for i in lst:
                file.write(i)
        return "Writing Successful"
    
    except IOError:
        print("Error in opening the file")

lst = ["1" , "2" , "3" , "4" , "abc"]
print(file_writing('new.txt',lst))
print(file_writing('old.txt',lst))
```
