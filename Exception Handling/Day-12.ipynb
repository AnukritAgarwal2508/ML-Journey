```python
def apply_discount(price, percent):
    if not 0 <= percent <= 100:
        raise ValueError(f"Discount must be 0-100, got {percent}")
    return price - (price * percent / 100)


# ---- Place 1: a web request. Show the user a message, keep serving. ----
def handle_web_request(price, percent):
    try:
        return {"status": "ok", "price": apply_discount(price, percent)}
    except ValueError as e:
        return {"status": "error", "message": str(e)}


# ---- Place 2: a batch job over 10,000 rows. Skip bad rows, keep going. ----
def process_batch(rows):
    results, skipped = [], 0
    for row in rows:
        try:
            results.append(apply_discount(row["price"], row["percent"]))
        except ValueError as e:
            print(f"Skipping bad row: {e}")
            skipped += 1
    print(f"Processed {len(results)}, skipped {skipped}")
    return results


# ---- Place 3: a startup config check. Bad config = refuse to run. ----
def load_pricing_config(percent):
    try:
        apply_discount(100, percent)
    except ValueError as e:
        raise SystemExit(f"Bad config, refusing to start: {e}")
```
