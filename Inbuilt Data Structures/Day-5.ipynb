```python
def max_min_sum(tuplee):
    max = float('-inf')
    min = float('+inf')
    sum = 0

    for num in tuplee:
        if num > max:
            max = num
        
        if num < min:
            min = num
        
        sum += num
        
    return max,min,sum


max,min,sum = max_min_sum((1,2,3,4,5))
print(max)
print(min)
print(sum)
```

```python
set1 = {3,4}
set2 = {3,4,5,6}

print(set1 | set2) #union
print(set1 & set2) #intersection
print(set1 - set2) #difference
print(set1 ^ set2)

#difference operation is not associative
print(set1 - set2 == set2 - set1)

#symmetric difference is associative
print(set1 ^ set2 == set2 ^ set1)
```

```python
#if A is subset of B for sure B will be superset of A
print(set1.issubset(set2) == set2.issuperset(set1))
```
