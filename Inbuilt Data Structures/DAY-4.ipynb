```python
def outer(x):
    def inner(y):
        return x * y
    
    ## returning a function
    return inner

# add contains the inner function now
add = outer(5)
print(add(10))
```

```python
def student_marks(science,maths):

    ''' we have defined a function inside a function here'''
    def average():
        return (science + maths)/2
    
    def grade():
        total_marks = science + maths
        if total_marks >= 90 and total_marks <= 100:
            return 'A+'
        
        elif total_marks >= 80 and total_marks <= 90:
            return 'B'
        
        elif total_marks >= 60 and total_marks <= 80:
            return 'C'
        
        else:
            return 'Fail'
    
    print("Average : ",average())
    print("Grade : ",grade())

student_marks(23,45)


```

```python
def decorator(func):

    #modified the orignal function with some changes
    #decorator recd the orignal function and transformed it to new one
    def new_greet():
        print("Function Starting")
        func()
        print("Function Ended")
    
    return new_greet

# @decorator
def greet():
    print("Hello World")


# greet()
#or we can do it this way as well 

new_greet = decorator(greet)
new_greet()
```

```python
def func_name_decorator(func):

    def new_function():
        print("Function name : ",func.__name__)
        func()

    
    return new_function

@func_name_decorator
def greet():
    print("Hello World")


greet()


```

```python
def decorator_printing_arguments(func):

    def wrapper(*args,**kwargs):
        print("Arguments : ",args)

        return func(*args,**kwargs)
    
    return wrapper


@decorator_printing_arguments
def add(a,b):
    return a + b


result = add(5 , 5)
print(result)
```

```python
def counting_decorator(func):

    def wrapper(*args,**kwargs):
        print("No of arguments passed : " , len(args))
        return func(*args,*kwargs)
    
    return wrapper

@counting_decorator
def new_add(a , b):
    return a + b


result = new_add(5 , 5)
print(result)

```

```python
def double_decorator(func):

    def double_wrapper(*args , **kwargs):
        return 2 * func(*args , **kwargs)
    
    return double_wrapper



@double_decorator
def square(a):
    return a ** 2

print(square(5))
```

```python
def function():
    x = 10

    print("First")

    yield x

    x = x + 5

    print("Second")

    yield x 

''' in order to use this function we have to use next() because yield keyword has made this a generator'''
# function()

gen = function()
print(next(gen))
print(next(gen))
```

```python
def fibonacci_sequence():
    first_value = 0
    second_value = 1

    yield first_value
    yield second_value

    while True:
        new_value = first_value + second_value
        yield new_value

        first_value = second_value
        second_value = new_value


fib = fibonacci_sequence()

for i in range(10):
    print(next(fib))
        
```

```python
def count_sum():

    sum = 0

    def inner_sum(number):
        
        nonlocal sum
        sum += number
        return sum
    
    return inner_sum

''' this will execute correctly because the closure is created only once'''
sum_tot = count_sum()
print(sum_tot(5))
print(sum_tot(5))
print(sum_tot(5))
print(sum_tot(5))


''' this will not because the closure is being created multiple times so the value sets back to 0 and further computations are done on it'''

# def count_sum(): sum = 0 def inner_sum(number): nonlocal sum sum += number return sum return inner_sum print(count_sum()(5)) print(count_sum()(5)) print(count_sum()(5)) print(count_sum()(5))


```

```python
def remember_name():
    names = []

    def add_name(name):
        nonlocal names
        names.append(name)
        return names
    
    return add_name

check = remember_name()

print(check('Anukrit'))
print(check('Aditya'))
print(check('Sreesh'))
```

```python
students = [

    {"name":"Anukrit","marks":92},

    {"name":"Rahul","marks":75},

    {"name":"Aditya","marks":88}

]

'''key in sorted function is used to tell the sorting condition when there too many factors on which sorting can be done'''
sorted_list = sorted(students,key = lambda students : students["marks"])
print(sorted_list)

''' reverse here is a condition in which we are telling the sorted function to sort in descending order'''
sorted_list_reversed = sorted(students,key = lambda students : students["marks"] , reverse = True)
print(sorted_list_reversed)
```
